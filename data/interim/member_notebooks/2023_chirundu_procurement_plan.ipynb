RE-EXCRATION OF PAGE 1 THAT IS CLEAN 

```python
import pdfplumber
import pandas as pd
import re

pdf_path = "2023-PROCUREMENT-PLAN.pdf"

# =========================================================
# 1. Read page 1
# =========================================================

with pdfplumber.open(pdf_path) as pdf:
    page = pdf.pages[0]
    words = page.extract_words()


# =========================================================
# 2. Define the useful columns
# =========================================================

columns = {
    "Category": (50, 113),
    "Procurement item": (196, 341),
    "Estimated Amount": (524, 548),
    "Source of Funds": (589, 637),
    "Procurement Method": (637, 661),
    "Publication": (661, 695),
    "date": (695, 730)
}


# =========================================================
# 3. Group words into physical rows
# =========================================================

rows = []

for word in words:

    top = word["top"]

    # Skip document title/header area
    if top < 165:
        continue

    matched_row = None

    for row in rows:
        if abs(row["top"] - top) <= 2:
            matched_row = row
            break

    if matched_row is None:
        matched_row = {
            "top": top,
            "words": []
        }

        rows.append(matched_row)

    matched_row["words"].append(word)


rows.sort(key=lambda x: x["top"])


# =========================================================
# 4. Extract the columns
# =========================================================

records = []

for row in rows:

    record = {}

    for column, (x_start, x_end) in columns.items():

        selected_words = [
            w["text"]
            for w in row["words"]
            if w["x0"] >= x_start and w["x0"] < x_end
        ]

        record[column] = " ".join(selected_words).strip()

    records.append(record)


df_2023 = pd.DataFrame(records)


# =========================================================
# 5. Remove rows that aren't procurement records
# =========================================================

valid_categories = [
    "goods",
    "works",
    "non consulting services",
    "consulting services"
]

df_2023["Category"] = (
    df_2023["Category"]
    .str.strip()
    .str.lower()
)

df_2023 = df_2023[
    df_2023["Category"].isin(valid_categories)
].copy()


# =========================================================
# 6. Clean Procurement item
# =========================================================

df_2023["Procurement item"] = (
    df_2023["Procurement item"]
    .str.replace(r"^0+", "", regex=True)
    .str.strip()
)


# =========================================================
# 7. Clean Estimated Amount
# =========================================================

df_2023["Estimated Amount"] = (
    df_2023["Estimated Amount"]
    .str.replace(",", "", regex=False)
    .str.replace(r"[^\d.]", "", regex=True)
)

df_2023["Estimated Amount"] = pd.to_numeric(
    df_2023["Estimated Amount"],
    errors="coerce"
)


# =========================================================
# 8. Clean Source of Funds
# =========================================================

def clean_source(value):

    if pd.isna(value):
        return ""

    value = str(value).strip()

    # Remove common OCR spacing
    value = re.sub(r"\s+", " ", value)

    # Local Resources
    if re.search(r"local\s*res", value, re.IGNORECASE):
        return "Local Resource"

    # Equalisation / Equalization / Equilization
    if re.search(
        r"equalisation|equalization|equilization",
        value,
        re.IGNORECASE
    ):
        return "Equalisation"

    # Year-based CDF
    match = re.search(
        r"(\d{4}/\d{4})\s*C?N?DF",
        value,
        re.IGNORECASE
    )

    if match:
        return f"{match.group(1)} CDF"

    # Normal CDF
    if re.search(r"\bC?N?DF\b", value, re.IGNORECASE):
        return "CDF"

    # Ward Fund
    if re.search(r"ward\s*fund", value, re.IGNORECASE):
        return "Ward Fund"

    return value


df_2023["Source of Funds"] = (
    df_2023["Source of Funds"]
    .apply(clean_source)
)


# =========================================================
# 9. Clean Procurement Method
# =========================================================

def clean_method(value):

    if pd.isna(value):
        return ""

    value = str(value).strip().lower()

    # Column overlap / OCR artifacts
    if "obn" in value:
        return "obn"

    if "db" in value:
        return "db"

    if "sb" in value:
        return "sb"

    return value


df_2023["Procurement Method"] = (
    df_2023["Procurement Method"]
    .apply(clean_method)
)


# =========================================================
# 10. Use AWARD as the final date
# =========================================================

df_2023["date"] = pd.to_datetime(
    df_2023["date"],
    format="%d-%m-%y",
    errors="coerce"
)


# =========================================================
# 11. Select final schema
# =========================================================

df_2023 = df_2023[
    [
        "Category",
        "Procurement item",
        "Estimated Amount",
        "Source of Funds",
        "Procurement Method",
        "date"
    ]
].reset_index(drop=True)


# =========================================================
# 12. Display result
# =========================================================

print("Total procurement records:", len(df_2023))

display(df_2023)
```
