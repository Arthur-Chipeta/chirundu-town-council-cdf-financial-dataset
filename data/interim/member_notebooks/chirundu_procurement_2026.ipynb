```python
df = pd.read_excel(
    "APP-CTC-2026.xls",
    sheet_name="Data",
    header=9
)
```

```python
for col in df.columns:
    print(repr(col))
```

```python
df_mined = df[
    [
        "Class",
        "Description",
        "Total",
        "Source of Funds",
        "Procurement Method",
        "Start "
    ]
]
```

```python
df_mined = df_mined.rename(columns={
    "Class": "Category",
    "Description": "Procurement item",
    "Total": "Estimated Amount",
    "Source of Funds": "Source of Funds",
    "Procurement Method": "Procurement Method",
    "Start ": "date"
})
```

Converting estimated amount to Numeric

```python
df_mined["Estimated Amount"] = pd.to_numeric(
    df_mined["Estimated Amount"],
    errors="coerce"
)
```

```python
df_mined["Estimated Amount"].head()
```

Coverting the date column

```python
df_mined["date"] = pd.to_datetime(
    df_mined["date"],
    errors="coerce"
)
```

```python
df_mined["date"].head()
```

Removal of completely empty Procurement records

```python
df_mined = df_mined.dropna(
    subset=["Category", "Procurement item"],
    how="all"
)
```

Reset the index

```python
df_mined = df_mined.reset_index(drop=True)
```

Checking the info

```python
df_mined.info()
```

Creation of the csv file.

```python
df_mined.to_csv(
    "chirundu_procurement_2026.csv",
    sep="|",
    index=False
)
```
