EXTRACTION OF PAGE 1

```python
import pdfplumber
import pandas as pd
import re

pdf_path = "2025-ANNUAL-PROCUREMENT-PLAN.pdf"

# Page 1
with pdfplumber.open(pdf_path) as pdf:
    page1 = pdf.pages[0]
    words1 = page1.extract_words()

# Column boundaries for Page 1
columns_page1 = {
    "Category": (50, 210),
    "Procurement item": (350, 566),
    "Estimated Amount": (790, 840)
}

# Group words by row
rows_page1 = []

for word in words1:

    # Ignore header information
    if word["top"] < 130:
        continue

    matched_row = None

    for row in rows_page1:
        if abs(row["top"] - word["top"]) <= 2:
            matched_row = row
            break

    if matched_row is None:
        matched_row = {
            "top": word["top"],
            "words": []
        }
        rows_page1.append(matched_row)

    matched_row["words"].append(word)

rows_page1.sort(key=lambda x: x["top"])

# Extract required fields
records_page1 = []

for row in rows_page1:

    record = {}

    for column, (x_start, x_end) in columns_page1.items():

        selected_words = [
            w["text"]
            for w in row["words"]
            if x_start <= w["x0"] < x_end
        ]

        record[column] = " ".join(selected_words).strip()

    records_page1.append(record)

df_page1 = pd.DataFrame(records_page1)

print(df_page1.head(10).to_string(index=False))
print("\nNumber of extracted rows:", len(df_page1))
```

```python
# Page 2
with pdfplumber.open(pdf_path) as pdf:
    page2 = pdf.pages[1]
    words2 = page2.extract_words()

# Column boundaries for Page 2
columns_page2 = {
    "Source of Funds": (60, 150),
    "Procurement Method": (240, 270),
    "date": (480, 520)
}

# Group words by row
rows_page2 = []

for word in words2:

    # Ignore header information
    if word["top"] < 130:
        continue

    matched_row = None

    for row in rows_page2:
        if abs(row["top"] - word["top"]) <= 2:
            matched_row = row
            break

    if matched_row is None:
        matched_row = {
            "top": word["top"],
            "words": []
        }
        rows_page2.append(matched_row)

    matched_row["words"].append(word)

rows_page2.sort(key=lambda x: x["top"])

# Extract fields
records_page2 = []

for row in rows_page2:

    record = {}

    for column, (x_start, x_end) in columns_page2.items():

        selected_words = [
            w["text"]
            for w in row["words"]
            if x_start <= w["x0"] < x_end
        ]

        record[column] = " ".join(selected_words).strip()

    records_page2.append(record)

df_page2 = pd.DataFrame(records_page2)

print(df_page2.head(10).to_string(index=False))
print("\nNumber of extracted rows:", len(df_page2))
```

REMOVE THE PAGE 2 FOOTER

```python
df_page2 = df_page2[
    df_page2["Source of Funds"].notna() &
    (df_page2["Source of Funds"].str.strip() != "")
].copy()

df_page2 = df_page2.reset_index(drop=True)

print("Number of procurement records:", len(df_page2))
```
